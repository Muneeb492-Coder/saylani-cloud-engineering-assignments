```python
import pandas as pd
import numpy as np

df = pd.read_csv("supermarket_sales_case_study.csv")
print(df.shape)   
df.head()
```

```python
df.info()
```

```python
print(df.isnull().sum())
```

```python
audit = pd.DataFrame({
    "missing_count": df.isnull().sum(),
    "missing_percent": (df.isnull().mean() * 100).round(2),
})
print(audit[audit.missing_count > 0])
print("exact duplicate rows:", df.duplicated().sum()) 
```

```python
df.duplicated().sum()
```

```python
df["Customer type"].fillna("unknown")
```

```python
df = df.drop_duplicates()
```

```python
print("Before exact duplicate rows:", df.duplicated().sum())   
df = df.drop_duplicates()
print("After cleaning exact duplicate rows:", df.duplicated().sum())  
```

```python
df["Customer type"] = df["Customer type"].fillna("unknown")
rating_median = df["Rating"].median()
df["Rating"] = df["Rating"].fillna(rating_median)
```

```python
df = df.dropna(subset=["Date"])
```

```python
print(df.isnull().sum())
```

```python
new_df = pd.DataFrame(df['Gender']) 
print(df["Gender"].unique())
print(df["Payment"].unique())
```

```python
df["Gender"] = df["Gender"].str.strip().str.title()

df["Payment"] = df["Payment"].str.strip().str.title()
```

```python
print(df["Gender"].unique())
print(df["Payment"].unique())
```

```python

df["time_clean"] = pd.to_datetime(df["Time"], errors="coerce")
df["date_clean"] = pd.to_datetime(df["Date"], errors="coerce")
```

```python
df.info()
```

```python
print(df["Quantity"].min())
```

```python
df["Quantity"] = df["Quantity"].abs()
```

```python
print(df["Invoice ID"].duplicated().sum())
```

```python
df = df.drop_duplicates(subset=["Invoice ID"], keep="last")
```
